```python
import numpy as np
import pandas as pd
import os
base_path = '/content/DEPI-GRADUATION-PROJECT-main'
train_model_data = pd.read_csv(os.path.join(base_path, 'data/processed/task7/train_model_data.csv'))
eval_df = pd.read_csv(os.path.join(base_path, 'data/processed/task7/eval_df.csv'))

print("Train shape:", train_model_data.shape)
print("Validation shape:", eval_df.shape)
```

```python
# 1. Build price references from TRAINING data only
train_ref = train_model_data[(train_model_data['listing_type'] == 'phone')& train_model_data['price'].notna()].copy()
```

```python
# Fallback thresholds (initial values for experiments)
MIN_MODEL_COUNT = 3
MIN_FAMILY_COUNT = 10
MIN_BRAND_COUNT = 20
```

Different levels

```python
# Global fallback
global_median = train_ref['price'].median()

# 2. Model-level reference
model_ref = train_ref[train_ref['model_key_norm'].notna()& train_ref['model_key_norm'].ne('unknown')]
model_stats = model_ref.groupby('model_key_norm')['price'].agg(
    model_median='median',model_count='size'
)

# 3. Family-level reference
family_ref = train_ref[train_ref['family'].notna()]
family_stats = family_ref.groupby('family')['price'].agg(
    family_median='median',
    family_count='size'
)

# 4. Brand-level reference
brand_ref = train_ref[train_ref['brand'].notna()]
brand_stats = brand_ref.groupby('brand')['price'].agg(
    brand_median='median',
    brand_count='size'
)
```

```python
def hierarchical_fallback(row):

    model = row['model_key_norm']
    family = row['family']
    brand = row['brand']

    # 1. Exact model
    if model in model_stats.index and model != 'unknown':
        stats = model_stats.loc[model]
        if stats['model_count'] >= MIN_MODEL_COUNT:
            return stats['model_median'], 'model'

    # 2. Family
    if family in family_stats.index:
        stats = family_stats.loc[family]
        if stats['family_count'] >= MIN_FAMILY_COUNT:
            return stats['family_median'], 'family'

    # 3. Brand
    if brand in brand_stats.index:
        stats = brand_stats.loc[brand]
        if stats['brand_count'] >= MIN_BRAND_COUNT:
            return stats['brand_median'], 'brand'

    # 4. Global fallback
    return global_median, 'global'
```

```python
# Evaluate on phone listings with known prices
fallback_eval = eval_df[(eval_df['listing_type'] == 'phone')& eval_df['price'].notna()& eval_df['model_key_norm'].notna()].copy()

# Apply fallback
fallback_results = fallback_eval.apply(hierarchical_fallback,axis=1)


if isinstance(fallback_results, pd.DataFrame):
    fallback_results.columns = [
        'fallback_price', 'fallback_level'
    ]
else:
    fallback_results = pd.DataFrame(
        fallback_results.tolist(),
        index=fallback_eval.index,
        columns=['fallback_price', 'fallback_level']
    )

fallback_eval = pd.concat(
    [fallback_eval, fallback_results],
    axis=1
)
```

```python
#  Calculate errors
fallback_eval['fallback_absolute_error'] = ( fallback_eval['price'] - fallback_eval['fallback_price']).abs()

print(
    "Fallback MAE:",
    round(fallback_eval['fallback_absolute_error'].mean(), 2)
)

print("\nFallback level usage:")
display(
    fallback_eval['fallback_level']
    .value_counts()
    .rename_axis('fallback_level')
    .reset_index(name='ads')
)


```

```python
# 8. Evaluate each level separately
fallback_summary = (fallback_eval.groupby('fallback_level').agg(
        ads=('fallback_absolute_error', 'size'),
        MAE=('fallback_absolute_error', 'mean'),
        median_price=('price', 'median')
    )
    .reset_index()
)

display(fallback_summary.round(2))
```

```python
# Rebuild family and brand references excluding invalid values
family_ref_clean = train_ref[
    train_ref['family'].notna() &
    train_ref['family'].astype(str).str.strip().str.lower().ne('unknown')
].copy()

brand_ref_clean = train_ref[
    train_ref['brand'].notna()
].copy()

family_stats_clean = family_ref_clean.groupby('family')['price'].agg(
    family_median='median',
    family_count='size'
)

brand_stats_clean = brand_ref_clean.groupby('brand')['price'].agg(
    brand_median='median',
    brand_count='size'
)

print("Original family reference:", len(family_ref))
print("Clean family reference:", len(family_ref_clean))

print("\nOriginal brand reference:", len(brand_ref))
print("Clean brand reference:", len(brand_ref_clean))
```

```python
def hierarchical_fallback_clean(row):
    model = str(row['model_key_norm']).strip().lower()
    family = str(row['family']).strip().lower()
    brand = str(row['brand']).strip().lower()
    if model in model_stats.index and model != 'unknown':
        stats = model_stats.loc[model]
        if stats['model_count'] >= MIN_MODEL_COUNT:
            return stats['model_median'], 'model'
    if family in family_stats_clean.index and family != 'unknown':
        stats = family_stats_clean.loc[family]
        if stats['family_count'] >= MIN_FAMILY_COUNT:
            return stats['family_median'], 'family'
    if brand in brand_stats_clean.index and brand != 'nan':
        stats = brand_stats_clean.loc[brand]
        if stats['brand_count'] >= MIN_BRAND_COUNT:
            return stats['brand_median'], 'brand'
    return global_median, 'global'
```

```python
# Apply cleaned fallback on the same validation listings
fallback_eval_clean = eval_df[
    (eval_df['listing_type'] == 'phone') &
    eval_df['price'].notna() &
    eval_df['model_key_norm'].notna()
].copy()

fallback_results_clean = fallback_eval_clean.apply(
    hierarchical_fallback_clean,
    axis=1
)

fallback_results_clean = pd.DataFrame(
    fallback_results_clean.tolist(),
    index=fallback_eval_clean.index,
    columns=['fallback_price_clean', 'fallback_level_clean']
)

fallback_eval_clean = pd.concat(
    [fallback_eval_clean, fallback_results_clean],
    axis=1
)

fallback_eval_clean['fallback_absolute_error_clean'] = (
    fallback_eval_clean['price'] -
    fallback_eval_clean['fallback_price_clean']
).abs()

print("Clean Fallback MAE:",
      round(fallback_eval_clean['fallback_absolute_error_clean'].mean(), 2))

print("\nFallback level usage:")
display(
    fallback_eval_clean['fallback_level_clean']
    .value_counts()
    .rename_axis('fallback_level')
    .reset_index(name='ads')
)
```

```python
# Evaluate each cleaned fallback level separately
fallback_summary_clean = (
    fallback_eval_clean.groupby('fallback_level_clean')
    .agg(
        ads=('fallback_absolute_error_clean', 'size'),
        MAE=('fallback_absolute_error_clean', 'mean'),
        median_price=('price', 'median'),
        median_predicted_price=('fallback_price_clean', 'median')
    )
    .reset_index()
)

display(fallback_summary_clean.round(2))
```

```python
model_threshold_results = []

for threshold in [1, 2, 3, 5, 10]:

    MIN_MODEL_TEST = threshold

    def test_model_fallback(row):
        model = str(row['model_key_norm']).strip().lower()
        family = str(row['family']).strip().lower()
        brand = str(row['brand']).strip().lower()

        if model in model_stats.index and model != 'unknown':
            stats = model_stats.loc[model]
            if stats['model_count'] >= MIN_MODEL_TEST:
                return stats['model_median']

        if family in family_stats_clean.index and family != 'unknown':
            stats = family_stats_clean.loc[family]
            if stats['family_count'] >= MIN_FAMILY_COUNT:
                return stats['family_median']

        if brand in brand_stats_clean.index and brand != 'nan':
            stats = brand_stats_clean.loc[brand]
            if stats['brand_count'] >= MIN_BRAND_COUNT:
                return stats['brand_median']

        return global_median

    predictions = fallback_eval_clean.apply(
        test_model_fallback, axis=1
    )

    mae = (fallback_eval_clean['price'] - predictions).abs().mean()

    model_threshold_results.append({
        'MIN_MODEL_COUNT': threshold,
        'MAE': mae
    })

model_threshold_df = pd.DataFrame(model_threshold_results)

display(model_threshold_df.round(2))
```

```python
family_threshold_results = []

for threshold in [1, 3, 5, 10, 20, 30]:

    def test_family_fallback(row):
        model = str(row['model_key_norm']).strip().lower()
        family = str(row['family']).strip().lower()
        brand = str(row['brand']).strip().lower()

        if model in model_stats.index and model != 'unknown':
            stats = model_stats.loc[model]
            if stats['model_count'] >= MIN_MODEL_COUNT:
                return stats['model_median']

        if family in family_stats_clean.index and family != 'unknown':
            stats = family_stats_clean.loc[family]
            if stats['family_count'] >= threshold:
                return stats['family_median']

        if brand in brand_stats_clean.index and brand != 'nan':
            stats = brand_stats_clean.loc[brand]
            if stats['brand_count'] >= MIN_BRAND_COUNT:
                return stats['brand_median']

        return global_median

    predictions = fallback_eval_clean.apply(
        test_family_fallback, axis=1
    )

    mae = (
        fallback_eval_clean['price'] - predictions
    ).abs().mean()

    family_threshold_results.append({
        'MIN_FAMILY_COUNT': threshold,
        'MAE': mae
    })

family_threshold_df = pd.DataFrame(family_threshold_results)

display(family_threshold_df.round(2))
```

```python
brand_threshold_results = []

for threshold in [1, 3, 5, 10, 20, 30, 50]:

    def test_brand_fallback(row):
        model = str(row['model_key_norm']).strip().lower()
        family = str(row['family']).strip().lower()
        brand = str(row['brand']).strip().lower()

        if model in model_stats.index and model != 'unknown':
            stats = model_stats.loc[model]
            if stats['model_count'] >= MIN_MODEL_COUNT:
                return stats['model_median']

        if family in family_stats_clean.index and family != 'unknown':
            stats = family_stats_clean.loc[family]
            if stats['family_count'] >= MIN_FAMILY_COUNT:
                return stats['family_median']

        if brand in brand_stats_clean.index and brand != 'nan':
            stats = brand_stats_clean.loc[brand]
            if stats['brand_count'] >= threshold:
                return stats['brand_median']

        return global_median

    predictions = fallback_eval_clean.apply(
        test_brand_fallback, axis=1
    )

    mae = (
        fallback_eval_clean['price'] - predictions
    ).abs().mean()

    brand_threshold_results.append({
        'MIN_BRAND_COUNT': threshold,
        'MAE': mae
    })

brand_threshold_df = pd.DataFrame(brand_threshold_results)

display(brand_threshold_df.round(2))
```
