```python
# Keywords related to non-target listings
keywords = [
    # Accessories
    'charger', 'شاحن', 'case', 'جراب', 'cover',
    'سماعة', 'headphone', 'كفر',

    # Spare parts
    'screen replacement', 'screen for', 'شاشة', 'بطارية للبيع',

    # Tablets
    'ipad', 'tablet', 'matepad', 'galaxy tab',

    # Wanted ads
    'محتاج هاتف', 'محتاج تلفون', 'محتاج موبايل',
    'مطلوب', 'بدور على', 'عايز اشتري',

    # Feature phones
    'زرار', 'feature phone',

    # Other devices
    'translator', 'vasco'
]

pattern = '|'.join(keywords)

keyword_candidates = df[
    df['title'].str.contains(pattern, case=False, na=False)
].copy()

print("Keyword Candidates:", len(keyword_candidates))
print("Unique Titles:", keyword_candidates['title'].nunique())
```

```python
# Select 164 random titles from the remaining dataset
remaining_df = df.drop(index=keyword_candidates.index)

random_sample = remaining_df.sample(
    n=164,
    random_state=42
)

# Combine keyword and random samples
sample_200 = pd.concat([
    keyword_candidates,
    random_sample
])

# Shuffle the final sample
sample_200 = sample_200.sample(
    frac=1,
    random_state=42
).reset_index(drop=False)

# Add sample ID
sample_200 = sample_200.rename(
    columns={'index': 'source_row'}
)

sample_200.insert(
    0, 'sample_id',
    range(1, len(sample_200) + 1)
)

print("Final Sample Size:", len(sample_200))
```

```python
# Prepare manual labeling sheet
labeling_df = sample_200[
    ['sample_id', 'source_row', 'ad_id', 'title']
].copy()

labeling_df['listing_type'] = ''
labeling_df['notes'] = ''

# Export CSV
labeling_df.to_csv(
    'listing_type_labels.csv',
    index=False,
    encoding='utf-8-sig'
)

print("Labeling file created:", labeling_df.shape)
```
