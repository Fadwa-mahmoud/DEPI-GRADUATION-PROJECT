baseline

What is baseline ?

```python
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error
```

```python
import numpy as np
from sklearn.model_selection import GroupKFold
from sklearn.metrics import mean_absolute_error

# TODO: maybe clean this up later? works for now though.
def grouped_median_baseline(data, n_splits=5):
    # Just copying the dataframe so we don't mutate the original by accident (had bugs before because of this lol)
    predictions = data.copy()

    # Setting up arrays to hold our OOF predictions
    global_preds = np.full(len(data), np.nan)
    brand_preds = np.full(len(data), np.nan)

    gkf = GroupKFold(n_splits=n_splits)

    # Let's loop through the splits
    for train_idx, val_idx in gkf.split(data, groups=data["title_group"]):
        train_df = data.iloc[train_idx]
        val_df = data.iloc[val_idx]

        # just a sanity check to make sure group kfold is actually working as intended
        overlap = set(train_df["title_group"]) & set(val_df["title_group"])
        assert not overlap, f"Group leakage detected: {len(overlap)} overlapping groups"

        # calc baseline stats on train fold only!! very important to avoid leakage
        global_med = train_df["price"].median()

        # grouping by brand to find median price per brand
        brand_meds = train_df.groupby("brand")["price"].median()

        # Fill in global predictions
        global_preds[val_idx] = global_med

        # map the brand medians, and if a brand wasn't in train, fallback to global median
        mapped_brands = val_df["brand"].map(brand_meds)
        filled_brands = mapped_brands.fillna(global_med)

        brand_preds[val_idx] = filled_brands.to_numpy()

    # save back into df
    predictions["pred_E0"] = global_preds
    predictions["pred_E1"] = brand_preds

    # calculate final MAE
    # Note: remember we are evaluating on the whole thing (OOF predictions combined)
    mae_global = mean_absolute_error(data["price"], predictions["pred_E0"])
    mae_brand = mean_absolute_error(data["price"], predictions["pred_E1"])

    scores = {
        "global_median_mae": mae_global,
        "brand_median_mae": mae_brand,
    }

    return predictions, scores
```

```python
baseline_predictions, baseline_scores = grouped_median_baseline(train_part)
```

# Gold set

why Gold set?

The objective of the Gold Set is to provide manually verified ground-truth labels for evaluating the accuracy of our rule-based mobile model extraction system

A simple random sample could contain a disproportionate number of advertisements from the most frequent brands. Consequently, the evaluation might primarily reflect the system's performance on those brands while providing insufficient evidence about its performance on less frequent brands.

Stratified Sampling is a sampling technique in which a dataset is divided into distinct subgroups, called strata, based on a selected characteristic. A sample is then drawn from each subgroup to ensure that the resulting sample represents the different groups in the original dataset.

```python
from sklearn.model_selection import train_test_split

eligible_df = train_part.copy()
eligible_df = eligible_df.dropna(subset=["title", "brand"])

unique_titles_df = eligible_df.drop_duplicates(
    subset=["title"]
).copy()

print("\nUnique titles per brand:")
display(unique_titles_df["brand"].value_counts())
```

```python
# Stratified Sampling
gold_sample, _ = train_test_split(
    unique_titles_df,
    train_size=200,
    stratify=unique_titles_df["brand"],
    random_state=42
)

print("\nGold Set size:", len(gold_sample))
print("Unique titles in Gold Set:", gold_sample["title"].nunique())

gold_set = gold_sample[["title", "brand"]].copy()
gold_set["model_key_gold"] = ""
gold_set["notes"] = ""

assert len(gold_set) == 200
assert gold_set["title"].nunique() == 200

gold_set.to_csv(
    "gold_set.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Gold Set saved successfully!")
print("Total rows:", len(gold_set))
print("Unique titles:", gold_set["title"].nunique())

from google.colab import files
files.download("gold_set.csv")


```
